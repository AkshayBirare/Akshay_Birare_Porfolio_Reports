```python
# Checking null values
print(df.isnull()) # Checking nulls
print(df.isnull().sum()) # Counting null per columns
```

```python
# Handling Null Values in Dataset
# Dropping null values, as replacing them might make difference in stastical analysis of dataset (mean, median, mode etc)
data_clean = df.dropna()
data_clean.isnull().sum()
```

```python
# Describing the basic info about the cleaned dataset
print(data_clean.shape)
print('_' * 49 ) # Line Seperator
print(data_clean.columns)
print('_' * 49 ) # Line Seperator
print(data_clean.dtypes)
print('_' * 49 ) # Line Seperator
print(data_clean.info())
print('_' * 49 ) # Line Seperator
print(data_clean.memory_usage())
print('_' * 49 )
```

```python
# Correct Summary Stastics
data_clean.describe()
```

```python
# Finding and handling outliers
import numpy as np
Q1 = data_clean.quantile(0.25)
Q3 = data_clean.quantile(0.72)
IQR = (Q3-Q1)/3
IQR # (Inter Quartile Range)
```

```python
Lower_Quartile = Q1 - 1.5*(IQR)
Upper_Quartile = Q3 + 1.5*(IQR)
Lower_Quartile , Upper_Quartile
```

```python
data_clean[data_clean > Upper_Quartile].count()
```

```python
data_clean[data_clean < Lower_Quartile].count()
```

```python
#Removing Outliers
data_without_outliers = data_clean[(data_clean <= Upper_Quartile) & (data_clean>= Lower_Quartile)]
data_without_outliers
```

```python
data_without_outliers.plot(kind='box' , figsize =(10,4))
```

```python
data_without_outliers.describe()
#Standard Deviation has decresed after treating outliers. Data became more accurate.
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
corr_matrix = data_without_outliers.corr()
corr_matrix
```

```python
#Creating Heatmap
import matplotlib.pyplot as plt
import seaborn as sns

# Set a modern, clean theme
sns.set_theme(style="whitegrid")

# Create heatmap with enhancements
plt.figure(figsize=(13, 9))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f", 
            linewidths=0.7, square=True, cbar_kws={"shrink": 0.75})

# Add title with formatting
plt.title("📊 Correlation Heatmap", fontsize=14, fontweight="bold")

# Improve axis labels for clarity
plt.xticks(rotation=45, fontsize=11)
plt.yticks(fontsize=11)

# Show the refined heatmap
plt.show()

#Variables are mostly positively correlated and highly dependent on each other.
```
